```python
# ============================================================
# 1. Import required libraries
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# 2. Load the cleaned dataset
# ============================================================

df = pd.read_csv("cleaned_cp_dataset.csv")

# ============================================================
# 3. Clean column names
# ============================================================

df.columns = df.columns.str.strip()

print("Columns in dataset:")
print(df.columns.tolist())

# ============================================================
# 4. Check required columns
# ============================================================

required_columns = ["formula", "T", "Cp"]

for column in required_columns:
    if column not in df.columns:
        raise ValueError(
            f"Required column '{column}' is missing.\n"
            f"Available columns are: {df.columns.tolist()}"
        )

# ============================================================
# 5. Create chemical family column if it doesn't exist
# ============================================================

if "family" not in df.columns:

    def classify_family(formula):
        formula = str(formula)

        if "Cl" in formula:
            return "Chlorides"
        elif "Br" in formula:
            return "Bromides"
        elif "F" in formula:
            return "Fluorides"
        elif "O" in formula:
            return "Oxides"
        elif "N" in formula:
            return "Nitrides"
        elif "S" in formula:
            return "Sulfides"
        elif "C" in formula:
            return "Carbides"
        else:
            return "Other"

    df["family"] = df["formula"].apply(classify_family)

# ============================================================
# 6. Remove missing values
# ============================================================

df = df.dropna(
    subset=["T", "Cp", "formula", "family"]
).copy()

# ============================================================
# 7. Make sure T and Cp are numeric
# ============================================================

df["T"] = pd.to_numeric(df["T"], errors="coerce")
df["Cp"] = pd.to_numeric(df["Cp"], errors="coerce")

df = df.dropna(
    subset=["T", "Cp"]
).copy()

# ============================================================
# 8. Sort data
# ============================================================

df = df.sort_values(
    by=["family", "formula", "T"]
)

# ============================================================
# 9. Set Seaborn style
# ============================================================

sns.set_theme(
    style="whitegrid",
    font_scale=1.1
)

# ============================================================
# 10. Create FacetGrid
# ============================================================

g = sns.FacetGrid(
    df,
    col="family",
    hue="formula",
    col_wrap=3,
    height=5,
    aspect=1.2,
    palette="tab10",
    sharex=False,
    sharey=False
)

# ============================================================
# 11. Plot Cp vs Temperature
# ============================================================

g.map(
    plt.plot,
    "T",
    "Cp",
    marker="o",
    linewidth=2,
    markersize=4
)

# ============================================================
# 12. Axis labels
# ============================================================

g.set_axis_labels(
    "Temperature (K)",
    "Heat Capacity, Cp (J/mol·K)"
)

# ============================================================
# 13. Facet titles
# ============================================================

g.set_titles(
    col_template="Chemical Family: {col_name}"
)

# ============================================================
# 14. Add legend
# ============================================================

g.add_legend(
    title="Material Formula"
)

# ============================================================
# 15. Overall title
# ============================================================

g.fig.suptitle(
    "Cp vs Temperature Across Chemical Families",
    fontsize=16,
    fontweight="bold",
    y=1.02
)

# ============================================================
# 16. Improve layout
# ============================================================

g.fig.tight_layout()

# ============================================================
# 17. Save figures BEFORE plt.show()
# ============================================================

png_file = "Cp_vs_T_FacetGrid.png"
pdf_file = "Cp_vs_T_FacetGrid.pdf"

g.fig.savefig(
    png_file,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

g.fig.savefig(
    pdf_file,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

print("✅ PNG saved as:", png_file)
print("✅ PDF saved as:", pdf_file)
print("✅ Successfully exported publication-ready figures at 300 DPI.")

# ============================================================
# 18. Display plot LAST
# ============================================================

plt.show()

# ============================================================
# 19. Close figure after displaying
# ============================================================

plt.close(g.fig)
```
